Task 20 — User-Defined Functions in Python

```python
def marks_summary(marks):
    if not marks:
        return {"total": 0, "average": None, "maximum": None, "minimum": None}
    return {
        "total": sum(marks),
        "average": sum(marks) / len(marks),
        "maximum": max(marks),
        "minimum": min(marks),
    }

marks = [78, 92, 85, 69, 88]
for key, value in marks_summary(marks).items():
    print(f"{key}: {value}")
```
