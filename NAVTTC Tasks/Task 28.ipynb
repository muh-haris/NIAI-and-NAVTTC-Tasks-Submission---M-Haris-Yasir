Task 28 — NumPy Create Array Using Python

```python
import numpy as np
arr_1d = np.array([1, 2, 3, 4])
arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
arr_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("1D array:\n", arr_1d)
print("2D array:\n", arr_2d)
print("3D array:\n", arr_3d)
```
