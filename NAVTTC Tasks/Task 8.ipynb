##Task 8 — Find the Area of a Triangle Using Heron's Formula
For side lengths a, b, and c, first calculate the semi-perimeter:

[ s = \frac{a+b+c}{2} ]

Then:

[ Area = \sqrt{s(s-a)(s-b)(s-c)} ]

```python
a = 5
b = 6
c = 7

s = (a + b + c) / 2
area = (s * (s - a) * (s - b) * (s - c)) ** 0.5

print("The area of the triangle is %.2f" % area)
```
