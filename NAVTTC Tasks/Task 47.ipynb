Task 47 — Pandas Data Cleaning Techniques

```python
import numpy as np
import pandas as pd

dirty = pd.DataFrame({
    "Name": [" Ali ", "Sara", "Sara", "Hamza"],
    "Age": ["22", "23", "23", "21"],
    "Score": [82, 91, 91, np.nan],
    "City": ["lahore", "Karachi", "Karachi", "LAHORE"]
})

print("Original:")
print(dirty)

clean = dirty.copy()

# Remove duplicate rows
clean = clean.drop_duplicates()

# Normalize text
clean["Name"] = clean["Name"].str.strip()
clean["City"] = clean["City"].str.strip().str.title()

# Convert data type
clean["Age"] = pd.to_numeric(clean["Age"], errors="coerce")

# Fill a missing numerical value with the column mean
clean["Score"] = clean["Score"].fillna(clean["Score"].mean())

print("\nCleaned:")
print(clean)
print("\nData types:")
print(clean.dtypes)
```
