Task 36 — NumPy Matrix / Array Joining

```python
import numpy as np
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

print("Concatenate by rows:\n", np.concatenate((a, b), axis=0))
print("Concatenate by columns:\n", np.concatenate((a, b), axis=1))
print("Vertical stack:\n", np.vstack((a, b)))
print("Horizontal stack:\n", np.hstack((a, b)))
```
