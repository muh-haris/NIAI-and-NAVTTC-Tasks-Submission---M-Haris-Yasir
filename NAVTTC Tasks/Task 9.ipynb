##Task 9 — Solve a Quadratic Equation
For:

[ ax^2 + bx + c = 0 ]

the discriminant is:

[ d = b^2 - 4ac ]

Using cmath.sqrt() allows the same code to handle both real and complex roots.

```python
import cmath

a = 1
b = 5
c = 6

d = (b ** 2) - (4 * a * c)

sol1 = (-b - cmath.sqrt(d)) / (2 * a)
sol2 = (-b + cmath.sqrt(d)) / (2 * a)

print("The solutions are {0} and {1}".format(sol1, sol2))
```
