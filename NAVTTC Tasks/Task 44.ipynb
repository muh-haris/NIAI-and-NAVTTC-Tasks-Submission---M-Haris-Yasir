Task 44 — Pandas DataFrame

```python
import pandas as pd
df = pd.DataFrame({
    "Name": ["Ali", "Sara", "Hamza", "Ayesha"],
    "Age": [22, 23, 21, 24],
    "Score": [82, 91, 76, 88]
})

print(df)
print("\nScore column:")
print(df["Score"])
print("\nSelected rows with loc:")
print(df.loc[1:2, ["Name", "Score"]])
```
