Task 40 — NumPy Random Technique

```python
import numpy as np
rng = np.random.default_rng(42)

random_integers = rng.integers(1, 11, size=10)
random_floats = rng.random(5)
normal_values = rng.normal(loc=100, scale=15, size=5)

print("Random integers:", random_integers)
print("Random floats:", np.round(random_floats, 4))
print("Normal samples:", np.round(normal_values, 2))
```
