Task 51: Implement Linear Regression

```python
from statistics import mean

x = [1, 2, 3, 4, 5]
y = [2, 4, 5, 4, 5]

x_bar = mean(x)
y_bar = mean(y)

slope = sum((xi - x_bar) * (yi - y_bar) for xi, yi in zip(x, y)) / \
        sum((xi - x_bar)**2 for xi in x)
intercept = y_bar - slope * x_bar
predictions = [intercept + slope * xi for xi in x]

print("Slope:", round(slope, 4))
print("Intercept:", round(intercept, 4))
print("Predictions:", [round(v, 4) for v in predictions])
```
