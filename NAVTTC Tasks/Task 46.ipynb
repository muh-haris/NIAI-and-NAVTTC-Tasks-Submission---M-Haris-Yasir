Task 46 — Pandas Data Analysis

```python
import pandas as pd
df = pd.DataFrame({
    "Age": [21, 22, 23, 24, 25],
    "StudyHours": [2, 3, 4, 5, 6],
    "Score": [60, 68, 75, 83, 90]
})

print("First rows:")
print(df.head())

print("\nDescriptive statistics:")
print(df.describe())

print("\nMean values:")
print(df.mean(numeric_only=True))
```
