Task 48 — Pandas Data Correlation

```python
import pandas as pd
df = pd.DataFrame({
    "StudyHours": [1, 2, 3, 4, 5, 6, 7],
    "Attendance": [60, 65, 70, 78, 82, 88, 92],
    "Score": [52, 57, 63, 69, 76, 83, 90]
})

correlations = df.corr(numeric_only=True)

print("Correlation matrix:")
print(correlations)

print("\nStudyHours vs Score:",
      round(correlations.loc["StudyHours", "Score"], 4))
```
