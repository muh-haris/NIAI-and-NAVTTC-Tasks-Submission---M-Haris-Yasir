Task 31 — NumPy Data Types

```python
import numpy as np
integer_array = np.array([1, 2, 3])
float_array = np.array([1, 2, 3], dtype=float)
string_array = np.array(["AI", "ML", "DL"])

print("Integer dtype:", integer_array.dtype)
print("Float dtype:", float_array.dtype)
print("String dtype:", string_array.dtype)

converted = integer_array.astype(np.float64)
print("Converted array:", converted)
print("Converted dtype:", converted.dtype)
```
