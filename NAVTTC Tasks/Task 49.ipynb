Task 49 — Perform Mean, Median, and Mode

```python
from statistics import mean, median, multimode

values = [99, 86, 87, 88, 111, 86, 103, 87, 94, 78, 77, 85, 86]

print("Values:", values)
print("Mean:", round(mean(values), 3))
print("Median:", median(values))
print("Mode(s):", multimode(values))
```
