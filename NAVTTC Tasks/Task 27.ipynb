Task 27 — User-Defined Exception Handling

```python
class InvalidAgeError(Exception):
    pass

def register_age(age):
    if age < 18:
        raise InvalidAgeError("Age must be at least 18.")
    if age > 100:
        raise InvalidAgeError("Age must not be greater than 100.")
    return f"Age {age} accepted."

for age in [25, 16, 105]:
    try:
        print(register_age(age))
    except InvalidAgeError as error:
        print("Custom exception:", error)
```
