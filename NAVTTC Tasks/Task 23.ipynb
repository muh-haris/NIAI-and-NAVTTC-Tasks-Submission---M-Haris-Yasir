Task 23 — String Data Type

```python
message = "  python for artificial intelligence  "

clean = message.strip()
title = clean.title()
words = clean.split()

print("Original:", repr(message))
print("Clean:", clean)
print("Title:", title)
print("First 6 characters:", clean[:6])
print("Words:", words)
print(f"Word count: {len(words)}")
```
