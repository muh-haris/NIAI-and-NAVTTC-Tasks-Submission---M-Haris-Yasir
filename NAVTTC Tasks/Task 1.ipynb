## Task 1 — Software Installation

The PDF asks for installation of:

- Anaconda3
- PyTorch
- TensorFlow 2.0
- VS Code
- PyCharm

```python
import sys
import importlib.util

print("Python Version:", sys.version.split()[0])

for library in ["torch", "tensorflow"]:
    if importlib.util.find_spec(library):
        module = __import__(library)
        print(f"{library}: {module.__version__}")
    else:
        print(f"{library}: Not installed")
```
