Task 37 — NumPy Array Splitting

```python
import numpy as np
arr = np.arange(1, 13)

three_parts = np.split(arr, 3)
print("Original:", arr)
print("Three equal parts:", three_parts)

matrix = arr.reshape(3, 4)
column_parts = np.hsplit(matrix, 2)
print("Matrix:\n", matrix)
print("Column split:", column_parts)
```
