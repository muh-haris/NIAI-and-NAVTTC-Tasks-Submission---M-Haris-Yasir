Task 26 — Exception Handling in Python

```python
def convert_and_divide(raw_value, divisor):
    try:
        number = int(raw_value)
        result = number / divisor
    except ValueError:
        return "ValueError: input is not a valid integer."
    except ZeroDivisionError:
        return "ZeroDivisionError: divisor cannot be zero."
    else:
        return f"Result = {result}"
    finally:
        print("Task 26 attempt completed.")

print(convert_and_divide("100", 4))
print(convert_and_divide("abc", 4))
print(convert_and_divide("100", 0))
```
