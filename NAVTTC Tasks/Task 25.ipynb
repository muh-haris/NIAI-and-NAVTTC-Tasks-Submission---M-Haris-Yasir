Task 25 — Dictionary Data Type

```python
student = {
    "name": "Hamza",
    "age": 22,
    "course": "AI"
}

student["week"] = 1
student["age"] = 23

print("Dictionary:", student)
print("Name:", student["name"])
print("Email:", student.get("email", "Not provided"))

for key, value in student.items():
    print(f"{key}: {value}")
```
