Work with:

Use **list comprehensions** (no regular for-loops!) to create and print each of the following:

1. A list of the **squares** of numbers 1 through 15: `[1, 4, 9, 16, ...]`
2. A list of numbers from 1–50 that are **divisible by 3 or 5** (but not both): `[3, 5, 6, 9, 10, ...]`

## 📝 Task 9 — Tuples: Immutability & Practical Uses

### Background
Tuples look like lists but use `()` instead of `[]`. They are **immutable** — you cannot add, remove, or change elements after creation. This makes them useful for fixed data like coordinates, RGB colours, or records. Tuples support indexing, slicing, and iteration just like lists.

You have coordinates of 5 cities:
```python

          ("Lahore",  31.55, 74.35),
          ("Islamabad", 33.72, 73.06),
          ("Peshawar", 34.01, 71.57),
          ("Quetta",  30.18, 66.99)]
```

- Sort the cities by **latitude** (ascending) and print the sorted list

> 💡 **Hint for sorting:** `sorted(cities, key=lambda c: c[1])` sorts by the second element of each tuple.


````python
# Part A
student = ("Alice", 22, "Computer Science", 3.8)

# YOUR CODE HERE
# Print name and GPA
print("Name:", student[0])
print("GPA:", student[3])

# Try to change name — catch TypeError
try:
    student[0] = "Bob"
except TypeError:
    print("Tuples are immutable!")

# Unpack into 4 variables and print each
name, age, major, gpa = student

print("Name:", name)
print("Age:", age)
print("Major:", major)
print("GPA:", gpa)

# Part B
cities = [("Karachi",   24.86, 67.01),
          ("Lahore",    31.55, 74.35),
          ("Islamabad", 33.72, 73.06),
          ("Peshawar",  34.01, 71.57),
          ("Quetta",    30.18, 66.99)]

# YOUR CODE HERE
# Loop and print each city's info
for city, lat, lon in cities:
    print(f"City: {city} | Lat: {lat} | Lon: {lon}")

# Sort by latitude and print
sorted_cities = sorted(cities, key=lambda c: c[1])
print(sorted_cities)
````

### Background
Just like list comprehensions, you can build dictionaries in one line:
```python
{key_expr: value_expr for item in iterable if condition}
```
Dictionaries are perfect for counting frequencies — a pattern you'll use constantly in data science.

### 🎯 Your Task

**Part A — Dictionary Comprehension:**
1. Create a dictionary mapping each number 1–10 to its **cube**: `{1: 1, 2: 8, 3: 27, ...}`
2. Create a dictionary mapping each number 1–20 to `"even"` or `"odd"`: `{1: 'odd', 2: 'even', ...}`

**Part B — Word Frequency Counter:**  
Count how many times each word appears in the text below, ignoring case. Print the result sorted by **frequency (descending)**.

```python
text = "to be or not to be that is the question to be is to think"
```
Expected top entries: `to: 4, be: 3, is: 2, ...`

> 💡 **Hint for Part B:** Start with `freq = {}`, then loop over `text.lower().split()`.  
> Use `freq[word] = freq.get(word, 0) + 1` to safely increment counts.  
> Sort with `sorted(freq.items(), key=lambda x: x[1], reverse=True)`.


````python
# Part A — Dictionary comprehensions
# 1. Cubes of 1 to 10
# YOUR CODE HERE
cubes = {n: n ** 3 for n in range(1, 11)}
print(cubes)

# 2. Even/Odd mapping for 1 to 20
# YOUR CODE HERE
even_odd = {n: "even" if n % 2 == 0 else "odd" for n in range(1, 21)}
print(even_odd)

# Part B — Word frequency counter
text = "to be or not to be that is the question to be is to think"
# YOUR CODE HERE

freq = {}

for word in text.lower().split():
    freq[word] = freq.get(word, 0) + 1

# Sort by frequency in descending order
sorted_freq = sorted(freq.items(), key=lambda x: x[1], reverse=True)

print(sorted_freq)
````

# 🔵 Section 5 — Sets


## 📝 Task 12 — Set Operations

### Background
A **set** is an **unordered collection of unique elements**. Duplicate values are automatically removed. Sets support mathematical operations: union (`|`), intersection (`&`), difference (`-`), symmetric difference (`^`).

| Operation | Symbol | Method | Meaning |
|-----------|--------|--------|---------|
| Union | `\|` | `.union()` | All elements from both sets |
| Intersection | `&` | `.intersection()` | Only elements in **both** |
| Difference | `-` | `.difference()` | Elements in A but **not** B |
| Symmetric Difference | `^` | `.symmetric_difference()` | Elements in either but **not both** |

### 🎯 Your Task
```python
python_students = {"Alice", "Bob", "Carol", "David", "Eve"}
java_students   = {"Bob", "David", "Frank", "Grace", "Alice"}
```

1. Find students enrolled in **both** courses (intersection)
2. Find students enrolled in **at least one** course (union)
3. Find students taking **Python but NOT Java** (difference)
4. Find students taking **exactly one** course (symmetric difference)
5. Check if `{"Bob", "Alice"}` is a **subset** of `python_students`
6. Add `"Henry"` to `python_students` and remove `"Carol"` from it
7. Create a list `[1, 2, 2, 3, 3, 3, 4, 4, 4, 4]` and use a set to print only the **unique values**

> 💡 **Hint:** Use `.issubset()` or the `<=` operator for subset check.


````python
python_students = {"Alice", "Bob", "Carol", "David", "Eve"}
java_students   = {"Bob", "David", "Frank", "Grace", "Alice"}

# YOUR CODE HERE
# 1. Intersection
both_courses = python_students & java_students
print("Both courses:", both_courses)

# 2. Union
at_least_one = python_students | java_students
print("At least one course:", at_least_one)

# 3. Difference (Python but NOT Java)
python_only = python_students - java_students
print("Python only:", python_only)

# 4. Symmetric difference
exactly_one = python_students ^ java_students
print("Exactly one course:", exactly_one)

# 5. Subset check: {"Bob", "Alice"} ⊆ python_students
is_subset = {"Bob", "Alice"}.issubset(python_students)
print("Is subset:", is_subset)

# 6. Add "Henry", remove "Carol"
python_students.add("Henry")
python_students.remove("Carol")
print("Updated Python students:", python_students)

# 7. Unique values from list
nums = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]

unique_values = set(nums)
print("Unique values:", unique_values)
````

## 📝 Task 15 — Vectorised Operations & Broadcasting

### Background
NumPy operations are applied **element-wise** to entire arrays without writing loops — this is called **vectorisation**. When arrays have compatible shapes, NumPy can **broadcast** smaller arrays across larger ones automatically.

### 🎯 Your Task

**Part A — Element-wise Operations:**

Compute and print (without loops):

4. Use the Boolean mask to extract only elements of `a` greater than 25

**Part B — Matrix Operations:**
```python
M = np.array([[1, 2], [3, 4]])
N = np.array([[5, 6], [7, 8]])
```
5. Element-wise multiplication `M * N`
6. Matrix (dot) product `M @ N` or `np.dot(M, N)`
7. Sum of all elements, sum along each row, sum along each column

> 💡 **Hint for Boolean masking:** `mask = a > 25` creates `[False False False True True]`. Then `a[mask]` extracts `[40 50]`.  
> 💡 **Hint for axis sums:** `M.sum(axis=1)` sums each row; `M.sum(axis=0)` sums each column.


````python
import numpy as np

a = np.array([10, 20, 30, 40, 50])
b = np.array([ 1,  2,  3,  4,  5])

# Part A
# YOUR CODE HERE
# 1. Element-wise +, -, *, /
print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("a / b =", a / b)

# 2. Square root of a
print("Square root of a:", np.sqrt(a))

# 3. Boolean mask: elements of a > 25
mask = a > 25
print("Mask:", mask)

# 4. Extract elements > 25 using mask
print("Elements greater than 25:", a[mask])

# Part B
M = np.array([[1, 2], [3, 4]])
N = np.array([[5, 6], [7, 8]])

# YOUR CODE HERE
print("Element-wise multiplication:")
print(M * N)

# 6. Matrix dot product
print("Matrix product:")
print(M @ N)

# 7. Total sum, row sums, column sums
print("Total sum:", M.sum())
print("Row sums:", M.sum(axis=1))
print("Column sums:", M.sum(axis=0))
````

## 📝 Task 18 — Capstone: Combined Data Structures Challenge

### Background
Real data science tasks require combining **all** the data structures you have learned. This final task tests your ability to choose the right structure for each job.

### 🎯 Your Task
You are given student records as a **list of dictionaries**:

```python
students = [
    {"name": "Alice",   "scores": [85, 90, 78], "city": "Lahore"},
    {"name": "Bob",     "scores": [60, 55, 70], "city": "Karachi"},
    {"name": "Carol",   "scores": [92, 88, 95], "city": "Islamabad"},
    {"name": "David",   "scores": [70, 65, 80], "city": "Lahore"},
    {"name": "Eve",     "scores": [50, 45, 55], "city": "Karachi"},
    {"name": "Frank",   "scores": [88, 92, 84], "city": "Islamabad"},
    {"name": "Grace",   "scores": [75, 80, 78], "city": "Lahore"},
]
```

Complete each of the following:

1. **Average score:** Add a key `"average"` to each student dict containing the average of their `scores` (round to 2 decimal places). Print all names and averages.

2. **Grade assignment:** Using a dictionary comprehension, create a dict `{name: grade}` where grade is:
   - `"A"` if average ≥ 85
   - `"B"` if average ≥ 70
   - `"C"` if average ≥ 55
   - `"F"` otherwise

3. **Top students:** Use a list comprehension to create a list of **names** of students with average ≥ 80.

4. **Unique cities:** Use a **set** to find all unique cities, then print them.

5. **NumPy statistics:** Convert all scores into a **2D NumPy array** (shape `7×3`). Compute and print:
   - Overall mean, min, max
   - Mean score per student (axis=1)
   - Mean score per test/column (axis=0)

6. **Pandas summary:** Create a **DataFrame** with columns `Name`, `Average`, `Grade`, `City`. Print it, then print the **average score per city** using `.groupby()`.

> 💡 **Hints:**
> - `sum(scores) / len(scores)` or `np.mean(scores)` for averages
> - `np.array([s["scores"] for s in students])` to build the 2D array
> - For the DataFrame, build a list of dicts or use parallel lists


````python
import numpy as np
import pandas as pd

students = [
    {"name": "Alice",  "scores": [85, 90, 78], "city": "Lahore"},
    {"name": "Bob",    "scores": [60, 55, 70], "city": "Karachi"},
    {"name": "Carol",  "scores": [92, 88, 95], "city": "Islamabad"},
    {"name": "David",  "scores": [70, 65, 80], "city": "Lahore"},
    {"name": "Eve",    "scores": [50, 45, 55], "city": "Karachi"},
    {"name": "Frank",  "scores": [88, 92, 84], "city": "Islamabad"},
    {"name": "Grace",  "scores": [75, 80, 78], "city": "Lahore"},
]

# YOUR CODE HERE — Step 1: Add "average" key to each student dict
for student in students:
    student["average"] = round(
        sum(student["scores"]) / len(student["scores"]), 2
    )

for student in students:
    print(student["name"], student["average"])


# Step 2: Dict comprehension {name: grade}
grades = {
    student["name"]:
        "A" if student["average"] >= 85
        else "B" if student["average"] >= 70
        else "C" if student["average"] >= 55
        else "F"
    for student in students
}

print("Grades:", grades)


# Step 3: List of names with average >= 80
top_students = [
    student["name"]
    for student in students
    if student["average"] >= 80
]

print("Top students:", top_students)


# Step 4: Set of unique cities
unique_cities = {student["city"] for student in students}

print("Unique cities:", unique_cities)


# Step 5: NumPy 2D array and statistics
scores_array = np.array([student["scores"] for student in students])

print("Scores array:")
print(scores_array)

print("Shape:", scores_array.shape)
print("Overall mean:", scores_array.mean())
print("Minimum score:", scores_array.min())
print("Maximum score:", scores_array.max())

print("Mean per student:")
print(scores_array.mean(axis=1))

print("Mean per test:")
print(scores_array.mean(axis=0))


# Step 6: Pandas DataFrame and groupby city
df = pd.DataFrame([
    {
        "Name": student["name"],
        "Average": student["average"],
        "Grade": grades[student["name"]],
        "City": student["city"]
    }
    for student in students
])

print(df)

print("Average score per city:")
print(df.groupby("City")["Average"].mean())
````
